7. Python Strings
 a. Slicing - Specifying the start and the end of characters 
 example: Hello World; 

```python
b = "HelloWorld"
print(b[2:5])
print(b[3:7])
print(b[1:8])
print(b[-8:-1])
print(b[-9:-1])
print(b[2:]) #index 2 and all other characters
print(b[:8])
```

b. Modifying strings
examples: .upper(), .lower(), .replace() , .strip()


```python
a = "                   Hello World"
print(a.upper())
print(a.lower())
print(a.replace("W", "V"))
```

c. Format Strings (f" ")

This is a type of string where we 'mix' numbers and text 
Example: I am 30 years old

To specify an f"" string, we add {} as placeholders for variables

```python
age = 30
name = "Sharon"

print(f"My name is {name} and I am {age} years old")
```

```python
price = 100
name = "Eldomatt"

print(f"The price of bread costs shs{price} at {name}")
```
