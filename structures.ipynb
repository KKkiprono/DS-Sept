# Indexing/slicing of lists
This is to specify a characters or items 

```python
fruits = ["apples", "banana", "cherry", "melon", "Oranges"]
print(fruits[1:3])
print(fruits[2:])
```

# Lists are changeable 
- You can replace, add or manipulate an item/character in a list

```python
fruits = ["apples", "banana", "cherry", "melon", "Oranges"]
fruits[1]= "Lemons"
print(fruits)
```

```python
fruits = ["apples", "banana", "cherry", "melon", "Oranges"]
fruits.insert(3, "berry")
print(fruits)
```

```python
fruits = ["apples", "banana", "cherry", "melon", "Oranges"]
fruits.remove("apples")
print(fruits)
```

```python
fruits = ["apples", "banana", "cherry", "melon", "Oranges"]
fruits.append("mangoes")
print(fruits[0])
print(fruits[1])
print(fruits[2])
print(fruits[3])
print(fruits[4])
print(fruits[5])
```

2. Tuples ()
A collection of ordered, unchangeable data. 
They allow duplication of values i.e cherry, banana, cherry , apple
We store data in ()

```python
fruits = ("apples", "bananas", "cherry")
print(fruits)
```

# Tuples are unchangeable 
We cannot change, add, remove or manipulate items/characters

```python
fruits = ("apples", "bananas", "cherry")
fruits[1] == "Melon" # Tuple did not change the order
print(fruits)
```

# Tuples slicing/indexing

```python
fruits = ("apples", "bananas", "cherry")
print(fruits[2])
print(fruits[:1])
print(fruits[1:])
```

3. Sets 
This is a collection/group of items/characters that are unordered and unchangeable 
Wes store data in callibraces{}
Does not allow duplicaions of items/characters

```python
names = {"Sharon", "Peter", "Albert", "Faith"}
print(names)
```

```python
#unordered
names = {"Sharon", "Peter", "Albert", "Faith"}
names.add("Fidel")
print(names)
```
