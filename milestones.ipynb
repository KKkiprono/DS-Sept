#Creating a shopping System 
Build a system that is to perfom the following functions:
1. View products
2. Choose a product
3. Check if the product exists
4. Enter quantity
5. Calculate cost
6. Buy another (Yes/No)

```python
# We create a shop

commodity = {
    "bread": 100,
    "milk": 40,
    "soda": 50,
    "crisps":30
}

print("My mini shopping system")

for product, price in commodity.items():
    print(product)
choice = input ("What would you like to buy?: ")

if choice in commodity:
    price = commodity[choice]
    quantity = int(input("How many would you like to buy?: "))
    cost = price * quantity
    print(f"You bought {quantity} packets of {choice} at {cost} shs")
else: 
    print("Sorry. Out of stock")
    
```

```python
# Banking system
# Deposit, withdraw, check balance, exit

balance = 1000

while True: 
    print("My Banking System")
    print("1.Deposit Money")
    print("2.Withdraw Money")
    print("3.Check balance")
    print("4.Exit")

    choice = input("Chooce an option: ")

    if choice == "1": 
        amount = float (input("Enter amount to deposit: "))
        balance = balance + amount
        print("Deposit successful")
        print(f"New balance is Kshs.{balance}")
        break

    elif choice "2":
        amount = float(input("Enter amount to withdraw: "))
        if amount <= balance:
            balance = balance - amount
            print("Withdrawal successful")
            print(f"New balance is Kshs.{balance}")
        else:
            print("Insufficient balance")
        break
    elif choice == "3":
        print(f"Your balance is Kshs.{balance}")
        break

    elif choice "4":
        print("Thank you for banking with us")
        break

    else:
        print("Inbvalid option. Choose option 1,2,3 or 4.")
```
