```python
# Arbitrary arguments (*args)
# *args allows a functions to receive any amount of positional arguments

#peviously
def add(a,b,c):
    print(a+b+c)
add(10,20,30)
```

```python
# using *args
def add(*args):
    print(args)
add(10,20,30,40,50)
```

```python
def show_students(*args):
    for student in args:
        print(student)
show_students("Abel", "Mary", "Brian", "Tony")
```

```python
def vehicles(*cars):
    for car in cars:
        print(car)
vehicles("Toyota", "Benz", "Chevrolet")
```

```python
# Normal arguments and *args

def student(name, *subjects):
    print(f"Student_name: {name}")
    print("Subjects enrolled: ")
    print("..................")

    for subject in subjects:
        print(subject)

student("Sammy", "Mathematics", "History", "Biology", "Computer", "English")
```

```python
#Shopping receipt
#Imagine a customer buys several items froma shop

def shopping_receipt(**shop):
    total = 0 
    for item in shop: 
        print(f"{item} : Kshs {shop[item]}")
        total = total + shop[item]
    print(f"Total amount: {total}")
shopping_receipt(Bread = 70, Milk = 60, Sugar = 120, Butter = 200)
```

 # Global Keyword
This is a keyword used inside a function when we want to change the value of a variable that was created outside that function. 

```python
balance = 1000
def withdraw():
    global balance 
    balance = balance - 200
withdraw()
print(balance)
```

```python
#Suppose we are admitting students in school. Everytime a student enters, the population increases by 1

students = 0
def registration():
    global students
    students = students + 1
registration()
registration()
registration()
print(f"I have {students} students in my class")
```

```python
school = "Highrise Academy"
def school_transfer():
    global school 
    school = "Fadhili Academy"
school_transfer()
print(school)
```

```python
price = 1000
def discount():
    global price 
    price = price - 200
discount()
print(price)
```

# Mini Banking System

Main Menu
1. Check balance
2. Deposit money
3. Withdraw money
4. Exit 

```python
# Mini banking system

balance = 5000
user_pin = "1234@k"
attempts = 0 

def check_balance(): 
    print(f"Your current balance is Kshs. {balance}")
def deposit():
    global balance
    amount = float(input("Enter the amount to deposit: "))
    if amount > 0:
        balance = balance + amount
        print(f"Kshs. {balance} deposited successfully.")
        print(f"Balance: {balance}")
    else: 
        print("Invalid amount. Enter an amount greater than zero")
def withdraw():
    global balance
    amount = float(input("Enter the amount to withdraw: "))
    if amount < 0:
        print("Invalid amount.")
    elif amount > balance:
        print("Insufficient balance. ")
    else:
        balance = balance - amount
        print(f"Kshs. {amount} withdrawn successfully.")
def menu():
    while True: 
        print("Welcome to the Main Menu")
        print("1. Check balance")
        print("2. Deposit money")
        print("3. Withdraw money")
        print("4. Exit")

        choice = input("Choose an option (1-4)")

        if choice == "1":
            check_balance()
        elif choice == "2":
            deposit()
        elif choice == "3":
            withdraw()
        elif choice == "4":
            print("Thank you for using the mini bank")
        else:
            print("Invalid option. Pick option (1-4)")

while attempts < 5:
    pin = input("Enter our user pin: ")
    if pin == user_pin: 
        print("Login successful")
        menu()
        break
    else: 
        print("Incorrect pin")
        attempts = attempts + 1
        remaining = 5 - attempts
        print(f"Attempts remaining: {remaining}")
        
```
