# Error handling
This allows a python program to handle errors during excecution without immeaditely crashing

```python
age = int(input("Enter age: "))
print(f"I am {age} years")
```

```python
try: 
    age = int(input("Enter your age"))
    print(f"You are {age} years old")
except ValueError:
    print("Invalid age. Please enter a number (int)")
```

```python
try:
    number1 = int(input("Enter first number: "))
    number2 = int(input("Enter second number: "))

    answer = number1/number2
    print(f"Answer: {answer}")

except ZeroDivisionError: 
    print("A number cannot be divided by zero")

except ValueError:
    print("Insert a number (int), an also not a float")
```

try - executes code that might raise an exception
except - handles a matching exception (fixing your own bugs)
else - runs if the try block finishes without an exception
finally - runs when leaving the try statement, whether an exception occured or not

```python
try:
    marks = float(input("Enter student marks: "))

except ValueError:
    print("Invalid marks")

else: 
    print(f"Marks entered: {marks}")

finally: 
    print("Marks processed successfully")
```

# Error handling - 
This handling of errors upon execution to avoid crashing
We are to use try, except, else, finally

```python
age = int(input("Enter your age: "))
print(f"You are {age} years old.")
```

```python
try:
    age = int(input("Enter your age: "))
    print(f"You are {age} years old.")
except ValueError:
    print("Invalid age! Please enter a number.")
```

```python
# Division by zero
try: 
    number1 = int(input("Enter first number: "))
    number2 = int(input("Enter second number: "))

    answer = number1 / number2
    print(f"Answer: {answer}")

except ZeroDivisionError:
    print("You cannot divide a number by zero")

except ValueError:
    print("Enter a a valid number; not a text")
```
