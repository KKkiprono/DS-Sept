- if block
Defines a condition is True, the code is executed
Terminate the line with a colon(:)
if ....... : 

- elif block (else-if)
Defines a previous condition as not True, then try this alternative 
elif: 

- else block
Defines a condition to provide an alternative 
Terminate the line with a colon(:)
else:



```python
age = 25
if age >= 18: 
    print("You are an adult")
    print("You are allowed to vote")
    print("You can acquire national ID")
```

```python
score = 90

if score >= 90:
    print("Excellent")
elif score >= 70: 
    print("Good")
elif score >= 50:
    print("Above average")
elif score <= 50: 
    print("Below average")
```

# Using if-else with User Input


```python
age = int(input("Enter your age: "))

if age >= 18: 
    print("You are eligible to vote")
else:
    print("You are an underage")
```

```python
correct_username = "admin"
correct_password = "4321"

username=input("Enter username: ")
password= input("Enter password: ")

#Make the decision
if username == correct_username and password == correct_password: 
    print("Access Granted!")
else: 
    print("Incorrect logins. Try again!")
```
