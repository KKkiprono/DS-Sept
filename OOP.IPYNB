# (self) Parameter
This refers to the particular object that is being currently used
It allows each object to store and access its own attributes and call its own methods

```python
class Students:
    def __init__(self, name, score):
        self.name = name
        self.score = score
    def display_results(self): #methods - simply because we are calling the action of displaying results
        print(f"{self.name} has scored {self.score} percent")
student1 = Students("Faith", 70)
student2 = Students("Abraham", 75)

student1.display_results() #student 1 is able to call his/her own results based on the method defined using self parameter
student2.display_results()
```

# OOP CLasses
1. Inheritance
This is a concept that allows one class to acquire attributes and methods from another class. 
Inheritance allows us to create a new class using features of an existing class, without redundancy 

Terms of Inheritance:
- Parent class = The class whose features are to be inherited
- Child class = The class that nherits features from the parent class
- Inheritance = The process of acquiring features from another class

```python
#Parent class
class Animals: 
    def eat(self):
        print("Most animals do eat")

#child class
class dogs(Animals):
    def bark(self):
        print("These dogs are barking so loud")

#creating objects
dog1 = dogs()

dog1.eat() #the common characteristic inherited from the parent
dog1.bark() #the specific characteristic of the child
```

Example 2: Student and teacher inheritance

```python
class human_beings: #parent class
    def __init__ (self, name, age):
        self.name = name
        self.age = age
    def dispaly_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
#child class
class student(human_beings):
    def study(self):
        print(f"{self.name} is studying Python Programming")

#child class
class teacher(human_beings):
    def teaching(self):
        print(f"{self.name} is teaching Python programming")

#create objects
student1 = student ("Ian", 22)
teacher1 = teacher ("Brian", 30)

#display output by calling the methods
student1.dispaly_info()
student1.study()
print()
teacher1.dispaly_info()
teacher1.teaching()
```

# Introducing the super() function
- Sometimes, a child needs additional information that is not included in the parent class
- We use the super() to call the parent's __init__method and initialize the inherited attributes

```python
class human_beings: 
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")

class students(human_beings):
    def __init__ (self, name, age, grade):
        super().__init__(name,age)
        self.grade = grade

    def display_info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Grade: {self.grade}")

person1 = human_beings("Jane", 30)
student1 = students ("Mary", 12, 9)

person1.display_info()
print()
student1.display_info()
```

# Polymorphism 
This is a concept that allows different objects to use the same method but perform different action

```python
class dog:
    def sound(self):
        print("The dog barks loudly")

class cat:
    def sound(self):
        print("The cat meows..")

class cow:
    def sound(self):
        print("The cow moows..")

dog1 = dog()
cat1 = cat()
cow1 = cow()

#visualize the output using methods
dog1.sound()
cat1.sound()
cow1.sound()
```

```python
#using loops n polymorphism
class dog:
    def sound(self):
        print("The dog barks loudly")

class cat:
    def sound(self):
        print("The cat meows..")

class cow:
    def sound(self):
        print("The cow moows..")

#create objects from the class
animals = [dog(), cat(), cow()]

#use loops
for animal in animals:
    animal.sound()
```

```python
#different vehicle with their parameters perfoming the same method but different actions
class car:
    def __init__ (self, brand, speed):
        self.brand = brand
        self.speed = speed
    def move(self):
        print(f"{self.brand} is driving at {self.speed} km/hr")

class plane:
    def __init__ (self, brand, speed):
        self.brand = brand
        self.speed = speed
    def move(self):
        print(f"{self.brand} is flying at {self.speed} knots")

class bicycle:
    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed
    def move (self):
        print(f"{self.brand} is riding at {self.speed} km/hr")

vehicles = [car("Toyota", 80), plane ("Boeing 747", 200), bicycle ("BMX", 50)] #create objects from class

for vehicle in vehicles:
    vehicle.move()
```

# Encapsulation 
This is about protecting data inside a class 
This involves keeping an object's data and the methods that work together in one class, while contrlling how the data is accessed or modified. 

```python
#Without using encapsulation: creating methods of Bank account
class BankAccount:
    def __init__ (self, name, balance):
        self.name = name
        self.balance = balance 

    def show_balance(self):
        print(f"Balance: Kshs {self.balance}")

account1 = BankAccount("Brian", 5000)
account1.show_balance()

account1.balance = 10000
account1.show_balance()

account1.balance = 150000
account1.show_balance()
```

```python
class BankAccount:
    def __init__ (self, name, balance):
        self.name = name # public identity
        self.__balance = balance #private data - we are securing the private information using double underscore

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount # total = initial + amount 
            print(f"The amount deposited is Ksh {amount}")

        else: 
            print("Invalid deposit amount")

    def show_balance(self):
        print(f"Balance is Kshs {self.__balance}")

account1 = BankAccount("Brian", 5000)
account1.deposit(2000)
account1.show_balance()
```

```python
# Student Marks Management
class Student:
    def __init__ (self, name, age , marks):
        self.name = name # public identity
        self._age = age # private information
        self.__marks = marks # protected information 

    def show_results (self):
        print(f"Student Name: {self.name}")
        print(f"Student Age: {self._age}")
        print(f"Student Marks: {self.__marks}")

    def update_results (self, marks):
        if 0 <= marks <= 100: 
            self.__marks = marks 
            print(f"Marks updated successfully")
        else: 
            print(f"Invalid Marks;  Enter between 0 and 100")

student1 = Student("Mary", 20, 60)
student2 = Student("Brian", 24, 75)

student1.update_results(66)
student1.show_results()
print()

student2.update_results(85)
student2.show_results()
```
