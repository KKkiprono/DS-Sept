# Loops 
This is simply repetition/iteration of a block of code. 
We have different ypes of loops; For loop & While loop 

For loop - A loop that repeats a block with a specific number of times 

```python
print("Hello World")
print("Hello World")
print("Hello World")
print("Hello World")
print("Hello World")
print("Hello World")
```

```python
# Use of range()

for i in range(6): 
    print("Hello World")
```

```python
for i in range(4):
    print(i)

# first loop = 0
# second loop = 1
# third loop = 2
# fourth loop = 3
```

```python
for number in range(5):
    print(number)
```

```python
# Start and stop 
for i in range(50, 69):
    print(i)

# starts at 50, stops at 69, but does not include 69
```

```python
# Counting of steps; start, stop, skip
# Selecting even numbers
for i in range(50, 69, 2):
    print(i)
```

```python
for i in range(50, 69, 3):
    print(i)
```

```python
for i in range(50, 40, -1):
    print(i)
```

```python
for i in range(50, 40, -1):
    print(i)
print("The end!!")
# The indented code is what gets repeated
```

```python
# Using For with if block
#Suppose we want to identify even numbers

for number in range(1, 11):
    if number % 2 == 0:
        print(number)
```

```python
# Using for in a list (looping through a list)
fruits = ["Apples", "Bananas", "Cherry", "Mangoes"]

for fruit in fruits:
    print(fruit)
```

```python
# Using for loop with dictionaries 

town = {
    "Shop 1" : {
        "name": "Laptop",
        "price": 50000,
        "quantity": 5
    },
    "Shop 2":{
        "name": "Mouse",
        "price": 1500,
        "quantity": 10
    },
    "Shop 3":{
        "name": "Keyboard",
        "price": 2000,
        "quantity": 7
    }
}
#Introduce key, value pairs in a dictionary
#Represent key = object, value= details and dictionary = town
#for key, value in dictionary.items():
for object, details in town.items(): 
    print(object)
    print("-------")
    print("name:", details["name"])
    print("price:", details["price"])
    print("quantity:", details ["quantity"])
    print()#skipping a line
```

# While loops
It is used to execute a set of statements as long as the condition is true
It iterates on infinite

```python
# To do an increment (i)

i = 1
while i < 6:
    print(i)
    i = i + 1
```

```python
count = 1

while count <= 15:
    print(count)
    count = count + 3
```

```python
# Countdown

count = 5

while count >= 1:
    print(count)
    count= count - 1
```

```python
password = ""
while password != "Python":
    password = input("Enter password: ")
print("Access Granted")
```

```python
# Guess a number

number = 0
while number != 10:
    number = int(input("Enter the number: "))
print("Correct Number")
```

```python
#combining if-else with while loop

password = "Python123"
attempts = 0

while attempts < 5:
    insert_password = input("Enter password: ")

    if insert_password == password:
        print("Access Granted")
        break #Stop iteration/loop upon success
    else: 
        attempts = attempts + 1
        print("Wrong Password, try again")

if attempts == 5:
    print("Maxiumum  attempts reached!. Access Denied")
```

```python
#combining if-else with while loop

password = "Python123"
attempts = 0

while attempts < 5:
    insert_password = input("Enter password: ")

    if insert_password == password:
        print("Access Granted")
        break #Stop iteration/loop upon success
    else: 
        attempts = attempts + 1
        print(f"Wrong Password, try again, {5- attempts} attempts remaining")

if attempts == 5:
    print("Maxiumum  attempts reached!. Access Denied")
```

```python
#combining if-else with while loop

password = "Python123"
attempts = 0

while attempts < 5:
    insert_password = input("Enter password: ")

    if insert_password == password:
        print("Access Granted")
        continue #Continue with iteration/loop upon success
    else: 
        attempts = attempts + 1
        print(f"Wrong Password, try again, {5- attempts} attempts remaining")

if attempts == 5:
    print("Maxiumum  attempts reached!. Access Denied")
```
