# Dictionaries
This is storing of data in Key:Value pairs
Data is ordered, changeable and no duplication
Key, value pairs are separated with a colon : (Key:Value)
Use of commas to terminate a line


```python
school = {
    "name" : "Eldohub", 
    "staff": "Gift",
    "est": 2019
    }

print(school)
```

```python
# Accessing information using keys
school = {
    "name" : "Eldohub", 
    "staff": "Gift",
    "est": 2019
    }

print(school["staff"])
print(school["est"])
```

```python
# Adding information
school = {
    "name" : "Eldohub", 
    "staff": "Gift",
    "est": 2019
    }
school["student"] = "Peter"
print(school)
```

```python
# Duplicates
school = {
    "name" : "Eldohub", 
    "staff": "Gift",
    "est": 2029, # Did not duplicate
    "est" : 2023
    }

print(school)
```

```python
# Duplicates
school = {
    "name" : "Eldohub", 
    "staff": "Gift",
    "est": 2029, 
    "year" : 2023
    }

print(school)
```
